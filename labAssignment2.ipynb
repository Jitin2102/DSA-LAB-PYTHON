### **Ques 2.	Write a program to find all prime numbers within a given range.**

```python
n =int(input("Enter a number: "))
if n <= 1:
    print("No prime number less than 1.")
else:
    for i in range(2, n+1):
        for j in range(2, int(i**0.5) + 1):
             if i % j == 0:
                break
        else:
            print(i,end=" ")
```

### **Ques 3.	Write a program to print ‘n' terms of Fibonacci series of numbers.**

```python
n =int(input("Enter a number: "))
a = 0
b = 1
for i in range(n):
    print(a,end=" ")
    c = a + b
    a = b
    b = c
```

### **Ques 4.	Write a program to compute the GCD of two numbers.**

```python
no1 = int(input("Enter first number: "))
no2 = int(input("Enter second number: "))
gcd = 1
for i in range(1, min(no1, no2) + 1):
    if no1 % i == 0 and no2 % i == 0:
        gcd = i
print("GCD of", no1, "and", no2, "is:", gcd)
```

### **Ques 6.	Write a Python program to implement a simple ATM machine. The program should:**  **Display options: Check Balance, Deposit, Withdraw, Exit.**   **Perform operations based on user input.**


```python
pincode = "5767"
balance = 1000
pin = input("Enter your 4-digit PIN: ")
if pin == pincode:
    print("PIN accepted. You can now access your account.")
    print("1. Check Balance")
    print("2. Withdraw Money")
    print("3. Deposit Money")
    print("4. Exit")
    while True:
        choice = input("Enter your choice (1-4): ")
        if choice == "1":
            print(f"Your balance is ₹{balance}.")
        elif choice == "2":
            amount = float(input("Enter amount to withdraw: ₹"))
            if amount <= balance:
                balance -= amount
                print(f"₹{amount} withdrawn successfully.")
            else:
                print("Insufficient balance.")
                break
        elif choice == "3":
            amount = float(input("Enter amount to deposit: ₹"))
            balance += amount
            print(f"₹{amount} deposited successfully.")
        elif choice == "4":
            print("Thank you for using the ATM.")
            break
        else:
            print("Invalid choice.")
else:
    print("Incorrect PIN. Access denied.")
```

### **Ques 6.	Write programs to print the following Patterns.**

### **Pattern 1**

```python
n = int(input("Enter a number: "))
for i in range(n,0,-1):
    print(" "*(n-i) + "*" * i)
```

### **Pattern 2**

```python
n = int(input("Enter a number: "))
for i in range(n, 0, -1):
    for j in range(n - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        print(chr(64 + i), end="")
    print()
```

### **Pattern 3**

```python
n = int(input("Enter a number: "))

for i in range(n + 1,-1,-1):
    for j in range(i, n + 1):
        print(j, end=" ")
    print()

for i in range(1, n+1, 1):
    for j in range(i, n + 1):
        print(j, end=" ")
    print()
```

### **Pattern 4**

```python
n = int(input("Enter a number: "))
for i in range(n, 0, -1):
    for j in range(n - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        print("*", end="")
    print()
```
