#### **This notebook is dedicated to Queue.**

```python
class Queue :
    def __init__(self):
        self.queue = []

    def enqueue(self,item):
        self.queue.append(item)

    def dequeue(self):
        if self.isEmpty():
            return "Queue is Empty"
        return self.queue.pop(0)

    def peek(self):
        if self.isEmpty():
            return "Queue is Empty"
        return self.queue[0]

    def isEmpty(self):
        return len(self.queue) == 0

    def size(self):
        return len(self.queue)
    def display(self):
        return self.queue
```

```python
q = Queue()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.enqueue(40)
print(q.peek())
q.dequeue()
print(q.peek())
```

#### **Reverse first k elements of a queue**

```python
def reverseFirstK(queue,k):
    if k <=0 or k > queue.size():
        return queue
    stack = []

    for _ in range(k):
        stack.append(queue.dequeue())

    while stack:
        queue.enqueue(stack.pop())

    size = queue.size()

    for _ in range(size - k):
        queue.enqueue(queue.dequeue())

    return queue
```

```python
q = Queue()
for val in [10, 20, 30, 40, 50]:
    q.enqueue(val)
print("Original Queue:", q.display())
reverseFirstK(q, 3)
print("Queue after reversing first 3 elements:", q.display())
```

#### **Checking Pallindrome using Queue**

```python
def isPallindrome(s):
    q = Queue()
    stack =[]

    for ch in s:
        q.enqueue(ch)
        stack.append(ch)
    while not q.isEmpty() :
        if q.dequeue() != stack.pop():
            return False
    return True
```

```python
s = "mad"
res = isPallindrome(s)
if res :
    print(f"{s} is a Pallindrome string.")
else :
    print(f"{s} is not a Pallindrome string.")
```
