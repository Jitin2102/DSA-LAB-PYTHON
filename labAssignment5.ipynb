#### **Ques 1.	Generate a list of squares for numbers from 1 to 10 using list comprehension.**

#### **Ques 4.	Given a list of numbers, use list comprehension to filter out the odd numbers.**

#### **Ques 5.	Given a list of words, create a list of their lengths using list comprehension.**

#### **Ques 6.	Given a string, extract all vowels using list comprehension.**

#### **Ques 8.	Given a list of words, create a new list with each word reversed using list comprehension.**

```python
words = ["Python", "Machine", "Learning", "Data", "Science"]
reversed_words = [word[::-1] for word in words]
print(f"Original: {words}")
print(f"Reversed: {reversed_words}")
```

#### **Ques 9.	Generate a list of prime numbers between 1 and 100 using list comprehension.**

```python
primes = [x for x in range(2, 101) if all(x % i != 0 for i in range(2, int(x**0.5) + 1))]
print(primes)
```

#### **Ques 10.	Convert a 2D matrix (list of lists) into a 1D list using list comprehension.**

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
flattened = [element for row in matrix for element in row]
print(f"2D Matrix: {matrix}")
print(f"1D Flattened List: {flattened}")
```

#### **Ques 17.	Write a program to remove all occurrences of a specific element from a tuple.**

```python
my_tuple = (10, 20, 30, 20, 40, 50, 20)
temp_list = list(my_tuple)
to_remove = 20
while to_remove in temp_list:
    temp_list.remove(to_remove)

res = tuple(temp_list)
print(f"No duplicate tuple: {res}")
```

#### **Ques 18.	Given a tuple, find the most frequently occurring element.**

```python
my_tuple = (10, 20, 30, 20, 40, 20, 50, 30, 30, 30)

frequency = {}
for item in my_tuple:
    frequency[item] = frequency.get(item, 0) + 1

most_frequent = None
max_count = 0

for key, value in frequency.items():
    if value > max_count:
        max_count = value
        most_frequent = key

print(f"Most Occuring Element: {most_frequent}")
```

#### **Ques 19.	Given a tuple of numbers, return the first element that does not repeat.**

```python
my_tuple = (4, 5, 1, 2, 0, 4, 1, 2)
counts = {}
for x in my_tuple:
    counts[x] = counts.get(x, 0) + 1
first_unique = None
for x in my_tuple:
    if counts[x] == 1:
        first_unique = x
        break

print(f"The first non-repeating element is: {first_unique}")
```

#### **Ques 20.	Given a tuple of integers and a target sum, find all pairs that add up to that sum.**

```python
def find_pairs_pointers(nums, target):
    nums = sorted(nums)
    left = 0
    right = len(nums) - 1
    pairs = []

    while left < right:
        current_sum = nums[left] + nums[right]
        if current_sum == target:
            pairs.append((nums[left], nums[right]))
            left += 1
            right -= 1
        elif current_sum < target:
            left += 1
        else:
            right -= 1
    return pairs

my_tuple = (2, 4, 3, 5, 7, 8, 9)
target_sum = 10
print(f"Pairs (Two-Pointer): {find_pairs_pointers(my_tuple, target_sum)}")
```
