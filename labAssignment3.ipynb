**Ques 10.	Duplicate Words: Identify and print all duplicate words in a string.**  
**o	Input: "this is is a test test"**  
**o	Output: {'is': 2, 'test': 2}**


```python
s = input("Enter a string: ")
dic = {}
if " " in s:
    l = s.split()
    for word in l:
        c = l.count(word)
        if c > 1:
            dic[word] = c
else:
    print("No duplicate words found")
print(dic)
```

**Ques 11.	String Compression: Compress a string using counts of repeated characters.**  
**o	Input: "aaabbcc"**  
**o	Output: "a3b2c2"**


```python
s = input("enter a string :")
result=""
if not s:
    print("enter a valid string")
else:
    i = 0
    while i < len(s):
        count = 1
        while i + 1 < len(s) and s[i] == s[i + 1]:
            count += 1
            i += 1
        result += s[i]
        if count > 1:
            result += str(count)
        i += 1
print(result)
```

**Ques12.	Longest Word: Find the longest word in a sentence.**  
**o	Input: "Find the longest word in this sentence"**  
**o	Output: "sentence"**


```python
s = input("Enter a sentence : ")
words = s.split()
ans = words[0]
le = len(words[0])
for i in range(len(words)):
    if le < len(words[i]):
        le = len(words[i])
        ans = words[i]

print(ans)
```

**Ques 13.	Anagram Check: Check if two strings are anagrams.**  
**An anagram of a string is another string that contains the same characters, only the order of characters can be different.**

```python
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")

if len(s1) != len(s2):
    print("Not anagrams")
else:
    if sorted(s1) == sorted(s2):
        print("Anagrams")
    else:
        print("Not anagrams")
```

**Ques 16.	Password Validator: Validate a password based on the following rules:**  
**o	Minimum 8 characters, at least one uppercase, one lowercase, one digit, and one special character.**  
**o	Input: "Password@123"**  
**o	Output: Valid**


```python
password = input("enter a password string: ")
if(len(password)<8):
    print("Invalid Password")

else:
    has_upper = any(ch.isupper() for ch in password)
    has_lower = any(ch.islower() for ch in password)
    has_digit = any(ch.isdigit() for ch in password)
    has_special = any(ch in "!@#$%^&*()-_=+[]{};:,.<>?/\\|" for ch in password)

    if has_upper and has_lower and has_digit and has_special:
        print("Valid Password")
    else:
        print("Invalid Password")
```

**Ques 18.	ASCII Conversion: Convert each character in a string to its ASCII value and vice versa.**  
**o	Input: "ABC"**  
**o	Output: [65, 66, 67]**  
**o	Input: [65, 66, 67]**  
**o	Output: "ABC"**


```python

text = "ABC"
ascii_values = [ord(ch) for ch in text]
print("String to ASCII:", ascii_values)

ascii_list = [65, 66, 67]
text_back = "".join(chr(num) for num in ascii_list)
print("ASCII to String:", text_back)
```
