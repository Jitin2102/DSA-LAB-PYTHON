## **Exercise from Exception Handling in Python** 

#### **Ques 1.	Write a Python program that prompts the user to input two numbers and raises a TypeError exception if the inputs are not numerical.**

```python
a = input("Enter the first number: ")
b = input("Enter the second number: ")

try:
    a = int(a)
    b = int(b)
except ValueError:
    raise TypeError("Inputs must be numerical values.")
```

#### **Ques 2.	Write a Python program that executes an operation on a list and handles an IndexError exception if the index is out of range.**

```python
list  = [1, 2, 3, 4, 5]
a = 5
try:
    print(list[a])
except IndexError:
    raise IndexError("Index out of range. Please enter a number between 0 and 4.")
```
