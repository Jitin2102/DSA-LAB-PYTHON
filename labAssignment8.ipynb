#### **Ques 1.	Write a recursive function to find the GCD of two numbers using Euclidean Algorithm.**

```python
def GCD(a, b):
    if b == 0:
        return a
    else:
        return GCD(b, a % b)
a ,b = map(int, input("Enter two numbers: ").split())
print("GCD of", a, "and", b, "is", GCD(a, b))
```

#### **Ques 2.	Write a recursive function to check if a given number n is prime.**

```python
def isprime(n,i=2):
    if n <= 1:
        return False
    if i * i > n:
        return True
    if n % i == 0:
        return False
    return isprime(n, i + 1)

num = int(input("Enter a number: "))

if isprime(num):
    print(num, "is a prime number.")
else:
    print(num, "is not a prime number.")
```

#### **Ques 3.	Implement a recursive function to count occurrences of a given character in a string.  count_occurrences("hello world", "o") → 2**


```python
def countOccurences(str,x,n):
    if not str:
        return 0
    if str[0] == x:
        return 1 + countOccurences(str[1:],x,n-1)
    return countOccurences(str[1:],x,n-1)
str = input("Enter the string: ")
print("String:", str)
x = input("Enter the character to count occurrences of: ")[0]
n = len(str)
print("Number of occurrences of", x, "in the string:", countOccurences(str, x, n))
```

#### **Ques 4.	Create a function calculator(operation) that defines four nested functions for addition, subtraction, multiplication, and division. It should return the appropriate function based on the operation.**
**add_func = calculator("add")**
**print(add_func(5, 3))**         


```python
def calculator(operation):
    def add(x, y):
        return x + y

    def subtract(x, y):
        return x - y

    def multiply(x, y):
        return x * y

    def divide(x, y):
        if y == 0:
            raise ValueError("Division by zero is not allowed")
        return x / y

    if operation == "add":
        return add
    elif operation == "subtract":
        return subtract
    elif operation == "multiply":
        return multiply
    elif operation == "divide":
        return divide
    else:
        raise ValueError("Invalid operation")

add_func = calculator("add")
print(add_func(5, 3))

sub_func = calculator("subtract")
print(sub_func(5, 3))

mul_func = calculator("multiply")
print(mul_func(5, 3))

div_func = calculator("divide")
print(div_func(10, 2))
```

#### **Ques 5.	Define a function power_function(base) that contains a nested function to calculate base raised to an exponent exp.**

```python
def power_function(base):
    def exponent(exp):
        return base ** exp

    return exponent

power_of_2 = power_function(2)
print(power_of_2(3))

power_of_5 = power_function(5)
print(power_of_5(4))
```

#### **Ques 6.	Write a function counter() that contains a nested function to keep track of the number of times it has been called.**
**c = counter()**  
**c()          # Output: Called 1 times**  
**c()          # Output: Called 2 times**


```python
def counter():
    a = 0

    def inner():
        nonlocal a
        a += 1
        print(f"Called {a} times")

    return inner
my_counter = counter()
my_counter()
my_counter()
```
